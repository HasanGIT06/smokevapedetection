```python
from ultralytics import YOLO
```

```python
model = YOLO("yolov8n.pt")
```

```python
model.train(
    data="/kaggle/input/datasets/leonardushasan/datafinaldl3/dataset_final/data.yaml",
    epochs=30,
    imgsz=512,
    batch=4,
    device=0
)
```

```python
model = YOLO("/kaggle/working/runs/detect/train/weights/best.pt")
metrics = model.val(data="/kaggle/input/datasets/leonardushasan/datafinaldl3/dataset_final/data.yaml")
```

```python
model.predict(source="/kaggle/input/datasets/leonardushasan/datafinaldl3/dataset_final/test/images", save=True, conf=0.25)
```
